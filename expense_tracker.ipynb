```python
import math

CURRENCY = "$"

# ---------- MODEL (logic) ----------
def parse_amount(text):
    amount = float(text)                       
    if not math.isfinite(amount) or amount <= 0:
        raise ValueError("Amount must be a positive number.")
    return amount
    
# ---------- VIEW (display) ----------
def show_welcome():
    print("===== EXPENSE TRACKER =====")
    print("\nEnter each expense amount. Type 'quit' to finish.\n")
    
def show_final_total(total, count):
    print("\n===========================")
    print(f"Expenses entered : {count}")
    print(f"FINAL TOTAL      : {CURRENCY}{total:.2f}")
    print("===========================")
    
# ---------- MAIN ----------
def main():
    total = 0.0     
    count = 0
    show_welcome()

    while True:
        entry = input("\nEnter expense amount (or 'quit'): ").strip()

        if entry.lower() == "quit":     
            break

        try:
            new_expense = parse_amount(entry)
        except ValueError:              
            print("Invalid data. Please enter a positive number like 100 or 49.50.")
            continue

        total += new_expense             
        count += 1
        print(f"Added {CURRENCY}{new_expense:.2f} | Running total: {CURRENCY}{total:.2f}")
        
    show_final_total(total, count)
    
if __name__ == "__main__":
    main()
```
